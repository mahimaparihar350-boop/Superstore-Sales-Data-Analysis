```python
# Sabse pehle 'kagglehub' nam ka tool Colab mein install karna padega
!pip install kagglehub pandas matplotlib seaborn

# 1. Apni Toolkit aur Data mangwana
import kagglehub
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns

# Aapka dhundha hua Kaggle ka official download code
path = kagglehub.dataset_download("ismatkhan121/supermarket-sales-dataset")
print("Data is folder mein aaya hai:", path)

# 2. Us folder ke andar jo .csv (table) file hai, usko dhundhna
files_in_folder = os.listdir(path)
csv_file_name = [f for f in files_in_folder if f.endswith('.csv')][0]
full_file_path = os.path.join(path, csv_file_name)

# 3. File ko 'df' (register) mein open karna
df = pd.read_csv(full_file_path)

print("\n✅ Asli Kaggle Data Load Ho Gaya! Shuru ki 5 lines:")
df.head()
```

```python
# STEP 3: Naye aur Sahi Graphs (Superstore Data ke hisaab se)
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.figure(figsize=(14, 5))

# Graph 1: Log konsa 'Ship Mode' (delivery type) sabse zyada use karte hain?
plt.subplot(1, 2, 1)
sns.countplot(x='Ship Mode', data=df, palette='Set2')
plt.title('1. Customer Shipping Mode Preference')
plt.xlabel('Ship Mode')
plt.ylabel('Count')

# Graph 2: Konsi 'Category' ka saaman sabse zyada bikta hai?
plt.subplot(1, 2, 2)
sns.countplot(x='Category', data=df, palette='Pastel1')
plt.title('2. Sales Count by Product Category')
plt.xlabel('Category')
plt.ylabel('Count')

plt.tight_layout()
plt.show()
```
